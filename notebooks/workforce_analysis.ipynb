# Workforce Analytics Case Study

## Objective

Analyze workforce productivity, utilization,
and employee health metrics
for a 10,000-employee organization over 2 years.

## Tools Used

- Python
- Pandas
- Matplotlib
- Power BI
- DAX

## Focus Areas

- Workforce utilization
- Productivity analysis
- Attrition indicators
- Burnout risk
- Operational efficiency

````python
import pandas as pd
import matplotlib.pyplot as plt
````

````python
# Load datasets

employee_df = pd.read_csv('../data/raw/employee_master.csv')

util_df = pd.read_csv('../data/raw/monthly_utilization.csv')
````

````python
# Merge datasets

df = pd.merge(
    util_df,
    employee_df,
    on='EmployeeID',
    how='left'
)
````

````python
print(df.shape)
df.head()
````

````python
# Dataset overview

df.info()
````

````python
# Missing values check

df.isnull().sum()
````

````python
# Basic statistical summary

df.describe()
````

````python
# Average utilization by department

dept_util = (
    df.groupby('Department')['UtilizationPct']
    .mean()
    .sort_values(ascending=False)
)

dept_util
````

````python
# Average utilization by department

dept_util = (
    df.groupby('Department')['Utilization_Pct']
    .mean()
    .sort_values(ascending=False)
)

dept_util
````

````python
plt.figure(figsize=(10,5))

dept_util.plot(kind='bar')

plt.title('Average Utilization by Department')

plt.ylabel('Utilization %')

plt.xticks(rotation=45)

plt.show()
````

````python
# Average leave hours by month

monthly_leave = (
    df.groupby('Month')['Leave_Hours']
    .mean()
)

monthly_leave
````

````python
# Convert Month column to datetime

df['Month'] = pd.to_datetime(df['Month'])

# Average leave hours by month

monthly_leave = (
    df.groupby('Month')['Leave_Hours']
    .mean()
)

# Plot line chart

plt.figure(figsize=(12,5))

monthly_leave.plot()

plt.title('Monthly Leave Trend')

plt.ylabel('Average Leave Hours')

plt.xticks(rotation=45)

plt.show()
````

````python
df.columns
````

````python
plt.figure(figsize=(8,6))

plt.scatter(
    df['Overtime_Hours'],
    df['PerformanceRating']
)

plt.xlabel('Overtime Hours')

plt.ylabel('Performance Rating')

plt.title('Overtime vs Performance')

plt.show()
````

````python
plt.figure(figsize=(8,6))

plt.scatter(
    df['Overtime_Hours'],
    df['PerformanceRating'],
    alpha=0.5
)

plt.xlabel('Overtime Hours')

plt.ylabel('Performance Rating')

plt.title('Overtime vs Performance')

plt.show()
````

````python
# Key Business Insights

## Workforce Productivity

- Certain departments demonstrated consistently higher utilization rates,
indicating strong productivity but also potential workload concentration.

- Workforce utilization imbalance suggests opportunities for improved resource allocation.

## Workforce Health & Attrition Indicators

- Elevated overtime patterns were observed across several employee groups,
which may indicate operational pressure and burnout risk.

- Increased leave trends during specific periods may reflect workforce stress concentration.

## Performance Analysis

- The relationship between overtime and performance indicates that
moderate overtime aligns with healthy productivity,
while excessive overtime may reduce workforce effectiveness.

## Strategic Recommendations

- Rebalance workload distribution across departments.

- Monitor high-risk employee groups with sustained overtime patterns.

- Improve workforce planning and capacity allocation.

- Implement employee wellness and retention initiatives for highly utilized teams.
````
